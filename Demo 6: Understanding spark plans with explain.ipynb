# Demo: Understanding spark exceution plans using explain()

```python
movies_df = spark.read.csv(
    "/Volumes/workspace/default/my_volume_data/movies.csv",
    header= True,
    inferSchema=True
)

movies_df.display()
```

```python
high_rated_df = movies_df.filter(
    (movies_df.averageRating >9) &
    (movies_df.numVotes > 1000000)
)
```

```python
high_rated_df.explain()
```

```python
high_rated_df.explain(True)
```
