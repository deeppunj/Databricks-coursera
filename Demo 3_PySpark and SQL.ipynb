```python
movies_df = spark.read.csv('/Volumes/workspace/default/my_volume_data/movies.csv', header = True, inferSchema = True)
```

```python
movies_df.display()
```

```python
movies_df.printSchema()
```

```python
movies_df.createOrReplaceTempView('movies') #this is a very important step for working with the data in SQL 
# this doesn't create anything but it actually just creates an instance of the same data, 
# but in reality Python and SQL will point to the same data movies_df
```

```python
%sql
SELECT primaryTitle, averageRating, numVotes
FROM movies
LIMIT 10;
```

```python
%sql
SELECT genres, Round(AVG(averageRating), 2) as avg_rating
FROM movies
GROUP BY genres
ORDER BY avg_rating DESC
LIMIT 10;
```

```python
from pyspark.sql.functions import avg, round

movies_df.groupBy('genres') \
    .agg(round(avg("averageRating"),2).alias('avg_rating')) \
    .orderBy('avg_rating', ascending = False) \
    .show(10)
```

```python
%sql
SELECT primaryTitle, averageRating
FROM movies
WHERE averageRating >= 7 
```
